````python
from transformers import AutoModel, AutoTokenizer
import numpy as np
import torchvision.transforms as T
from PIL import Image
from torchvision.transforms.functional import InterpolationMode

IMAGENET_MEAN = (0.485, 0.456, 0.406)
IMAGENET_STD = (0.229, 0.224, 0.225)

def build_transform(input_size):
    MEAN, STD = IMAGENET_MEAN, IMAGENET_STD
    transform = T.Compose([
        T.Lambda(lambda img: img.convert('RGB') if img.mode != 'RGB' else img),
        T.Resize((input_size, input_size), interpolation=InterpolationMode.BICUBIC),
        T.ToTensor(),
        T.Normalize(mean=MEAN, std=STD)
    ])
    return transform

def find_closest_aspect_ratio(aspect_ratio, target_ratios, width, height, image_size):
    best_ratio_diff = float('inf')
    best_ratio = (1, 1)
    area = width * height
    for ratio in target_ratios:
        target_aspect_ratio = ratio[0] / ratio[1]
        ratio_diff = abs(aspect_ratio - target_aspect_ratio)
        if ratio_diff < best_ratio_diff:
            best_ratio_diff = ratio_diff
            best_ratio = ratio
        elif ratio_diff == best_ratio_diff:
            if area > 0.5 * image_size * image_size * ratio[0] * ratio[1]:
                best_ratio = ratio
    return best_ratio

def dynamic_preprocess(image, min_num=1, max_num=12, image_size=448, use_thumbnail=False):

    orig_width, orig_height = image.size
    aspect_ratio = orig_width / orig_height

    # calculate the existing image aspect ratio
    target_ratios = set(
        (i, j) for n in range(min_num, max_num + 1) for i in range(1, n + 1) for j in range(1, n + 1) if
        i * j <= max_num and i * j >= min_num)
    target_ratios = sorted(target_ratios, key=lambda x: x[0] * x[1])

    # find the closest aspect ratio to the target
    target_aspect_ratio = find_closest_aspect_ratio(
        aspect_ratio, target_ratios, orig_width, orig_height, image_size)

    # calculate the target width and height
    target_width = image_size * target_aspect_ratio[0]
    target_height = image_size * target_aspect_ratio[1]
    blocks = target_aspect_ratio[0] * target_aspect_ratio[1]

    # resize the image
    resized_img = image.resize((target_width, target_height))
    processed_images = []
    for i in range(blocks):
        box = (
            (i % (target_width // image_size)) * image_size,
            (i // (target_width // image_size)) * image_size,
            ((i % (target_width // image_size)) + 1) * image_size,
            ((i // (target_width // image_size)) + 1) * image_size
        )
        # split the image
        split_img = resized_img.crop(box)
        processed_images.append(split_img)
    assert len(processed_images) == blocks
    if use_thumbnail and len(processed_images) != 1:
        thumbnail_img = image.resize((image_size, image_size))
        processed_images.append(thumbnail_img)
    return processed_images

def load_image(image_file, input_size=448, max_num=12):
    image = Image.open(image_file).convert('RGB')
    transform = build_transform(input_size=input_size)
    images = dynamic_preprocess(image, image_size=input_size, use_thumbnail=True, max_num=max_num)
    pixel_values = [transform(image) for image in images]
    pixel_values = torch.stack(pixel_values)
    return pixel_values
````

#### Inference for table image + TSR output (HTML) to LaTeX code

````python
import os
from tqdm import tqdm
import torch

def process_images_and_html_to_latex(src_text_file: str, image_dir: str, output_file: str, model, tokenizer, generation_config):
    """Process images and their HTML representation to generate LaTeX code with progress bar"""
    
    with open(src_text_file, "r") as file:
        lines = file.readlines()
    
    data_entries = []
    i = 0
    while i < len(lines):
        line = lines[i].strip()
        if line and '\t' in line:  # Check if line contains a tab (image_id and HTML)
            parts = line.split('\t')
            if len(parts) >= 2:
                image_id = parts[0]
                html_representation = parts[1]
                data_entries.append((image_id, html_representation))
        i += 1
    
    results_buffer = []
    
    for image_id, html_representation in tqdm(data_entries, desc="Processing images", unit="image"):
        image_path = os.path.join(image_dir, f"{image_id}")
        
        if not os.path.exists(image_path):
            print(f"Warning: {image_path} not found, skipping.")
            continue
        
        pixel_values = load_image(image_path, max_num=12).to(torch.bfloat16).cuda()
        
        prompt = f"""<image> Given a table image and its corresponding HTML representation, generate executable LaTeX code that accurately represents the table's structure. Use the table image to infer visual elements such as grid lines (horizontal and vertical), row and column merging (rowspan and colspan), and text alignment within cells (left, center, or right). Use the provided HTML representation as the structural reference to ensure the number of rows and columns match, rowspan and colspan attributes are correctly converted, and cell contents are placeholders ("CELL") regardless of the actual text. Generate valid executable LaTeX code using \\begin{{tabular}} with column specifications including vertical lines (e.g., {{|l|c|r|}}), horizontal lines (\\hline) at the top, bottom, and between rows (based on the image), and proper handling of merged cells using \\multirow and \\multicolumn. Inputs: (1) Table Image, (2) HTML Table Representation: {html_representation}"""
        
        # Get the model's response (generated LaTeX code)
        response = model.chat(tokenizer, pixel_values, prompt, generation_config)
        
        results_buffer.append({"image_filename": image_id, "latex_output": response})
        
        if len(results_buffer) >= 50:
            write_results_to_file(results_buffer, output_file)
    
    if results_buffer:
        write_results_to_file(results_buffer, output_file)
    
    print("LaTeX generation completed for all images.")

def write_results_to_file(results_buffer, output_file: str):
    """Write current buffer to file and clear it"""
    with open(output_file, 'a', encoding='utf-8') as f:
        for result in results_buffer:
            f.write(f"{result['image_filename']}\t{result['latex_output']}\n\n")
    
    results_buffer.clear()

src_text_file = "/iitjhome/unitable/table_structure_predictions.txt"
image_dir = "/scratch/data/tab2tex/tablev1.2/images/square"
output_file = "/iitjhome/unitable/ZS_internvl_HTML_TableImage_placeholder.txt"

process_images_and_html_to_latex(src_text_file, image_dir, output_file, model, tokenizer, generation_config)
````

````python
import os
import json
import random
from PIL import Image

image_dir = "/scratch/data/tab2tex/tablev1.2/images/square/"
src_train_path = "/scratch/data/tab2tex/tablev1.2/structure/src-train.txt"
tgt_train_path = "/scratch/data/tab2tex/tablev1.2/structure/tgt-train.txt"
output_30k_jsonl_path = "/scratch/data/tab2tex/train_dataset_ImageOnly_FAT_30k.jsonl"
output_remaining_jsonl_path = "/scratch/data/tab2tex/train_dataset_ImageOnly_FAT_remaining.jsonl"

# Number of images for the first file
NUM_IMAGES_FIRST_FILE = 30000

with open(src_train_path, "r") as src_file:
    image_paths = [line.strip() for line in src_file.readlines()]

with open(tgt_train_path, "r") as tgt_file:
    ground_truths = [line.strip() for line in tgt_file.readlines()]

assert len(image_paths) == len(ground_truths), "Mismatch between src-train.txt and tgt-train.txt!"

paired_data = list(zip(image_paths, ground_truths))
total_samples = len(paired_data)

print(f"Total dataset size: {total_samples} samples")

# Randomly select 30,000 samples for the first file
if total_samples > NUM_IMAGES_FIRST_FILE:
    first_file_samples = random.sample(paired_data, NUM_IMAGES_FIRST_FILE)
    
    # Get the remaining samples (around 13,138)
    remaining_samples = [item for item in paired_data if item not in first_file_samples]
    
    print(f"Selected {len(first_file_samples)} samples for first file")
    print(f"Selected {len(remaining_samples)} samples for second file")
else:
    print(f"Warning: Requested {NUM_IMAGES_FIRST_FILE} images but only {total_samples} are available")
    first_file_samples = paired_data
    remaining_samples = []

base_prompt = """<image>

Given a table image, generate executable LaTeX code that accurately reconstructs the table's structure.

### **Guidelines:**
1. **Visual Information (from Image)**:
   - Detect grid structure (horizontal and vertical lines).
   - Identify row and column merging (e.g., spans across rows or columns).
   - Determine text alignment (left, center, right).
   - Ensure the correct number of rows and columns.
   - Treat all cell contents as **"CELL"** placeholders, regardless of actual text.

2. **Expected LaTeX Format**:
   - Start with `{ alignment }`Use appropriate column alignments (e.g., `{|l|c|r|}`, `{l l}`) based on the image.
   - Add `\\hline` for horizontal lines where visible.
   - Use `\\multirow` and `\\multicolumn` for merged cells.

**Expected Output**:
Generate valid executable LaTeX code that precisely represents the table.
"""


def process_samples(samples, output_path):
    entries = []
    sample_count = len(samples)
    
    for idx, (image_path, latex) in enumerate(samples):
        full_image_path = os.path.join(image_dir, image_path)
        
        if not os.path.exists(full_image_path):
            print(f"Warning: Image not found: {full_image_path}")
            continue
            
        try:
            with Image.open(full_image_path) as img:
                width, height = img.size
        except Exception as e:
            print(f"Error opening image {full_image_path}: {e}")
            continue
            
        entry = {
            "id": idx,
            "image": image_path,  # Relative path to the image
            "width": width,
            "height": height,
            "conversations": [
                {
                    "from": "human",
                    "value": base_prompt
                },
                {
                    "from": "gpt",
                    "value": latex
                }
            ]
        }
        entries.append(entry)
        
        if (idx + 1) % 1000 == 0 or idx + 1 == sample_count:
            print(f"Processed {idx + 1}/{sample_count} images.")
    
    with open(output_path, "w") as output_file:
        for entry in entries:
            output_file.write(json.dumps(entry) + "\n")
    
    print(f"JSONL file created at: {output_path} with {len(entries)} entries")
    return len(entries)

print("Processing 30k samples...")
processed_30k = process_samples(first_file_samples, output_30k_jsonl_path)

if remaining_samples:
    print("Processing remaining samples...")
    processed_remaining = process_samples(remaining_samples, output_remaining_jsonl_path)
    print(f"Total processed: {processed_30k + processed_remaining} out of {total_samples}")
else:
    print("No remaining samples to process")
    print(f"Total processed: {processed_30k} out of {total_samples}")
````

````python
import os
import torch
from transformers import AutoModel, AutoTokenizer

def process_single_image(image_path, model, tokenizer, generation_config):
    """Process a single image to generate LaTeX code using vision-only prompt"""
    
    if not os.path.exists(image_path):
        print(f"Error: Image {image_path} not found.")
        return None
    
    device = next(model.parameters()).device
    pixel_values = load_image(image_path, max_num=12).to(device, torch.bfloat16)
    
    prompt = """<image>
    
Given a table image, generate executable LaTeX code that accurately reconstructs the table's structure.

### **Guidelines:**
1. **Visual Information (from Image)**:
   - Detect grid structure (horizontal and vertical lines).
   - Identify row and column merging (e.g., spans across rows or columns).
   - Determine text alignment (left, center, right).
   - Ensure the correct number of rows and columns.
   - Treat all cell contents as **"CELL"** placeholders, regardless of actual text.

2. **Expected LaTeX Format**:
   - Start with `{ alignment }` Use appropriate column alignments (e.g., `{|l|c|r|}`, `{l l}`) based on the image.
   - Add `\\hline` for horizontal lines where visible.
   - Use `\\multirow` and `\\multicolumn` for merged cells.

**Expected Output**: Generate valid executable LaTeX code that precisely represents the table.
"""
    
    response = model.chat(tokenizer, pixel_values, prompt, generation_config)
    
    return response
````

````python
import os
import json
from PIL import Image

image_dir = "/scratch/data/tab2tex/tablev1.2/images/square/"
rowcol_path = "/scratch/data/tab2tex/table_counts_train_cell_based_with_filenames.txt"
latex_pred_path = "/unitable/FT_internvl_M2_inference_13k.txt"
groundtruth_latex_jsonl_path = "/scratch/data/tab2tex/train_dataset_ImageOnly_FAT_remaining.jsonl"  # Updated path
output_jsonl_path = "/scratch/data/tab2tex/train_dataset_Image_row_col_latex_FAT_CoTv2.jsonl"

rowcol_info = {}
with open(rowcol_path, "r") as f:
    for line in f:
        parts = line.strip().split()
        if len(parts) >= 5:
            filename = parts[0]
            rows = int(parts[2].strip(","))
            cols = int(parts[4])
            rowcol_info[filename] = {"rows": rows, "cols": cols}

latex_preds = {}
with open(latex_pred_path, "r") as f:
    for line in f:
        if '\t' not in line:
            continue
        filename, latex = line.strip().split('\t', 1)
        latex_preds[filename] = latex

groundtruth_latex = {}
with open(groundtruth_latex_jsonl_path, "r") as f:
    for line in f:
        entry = json.loads(line)
        filename = entry["image"]
        gt_latex = entry["conversations"][-1]["value"] 
        groundtruth_latex[filename] = gt_latex

def create_prompt(rows, cols, initial_latex, groundtruth_latex):
    return f"""<image>

You are given a table image, an initial LaTeX prediction, and the correct number of rows and columns in the table. Your task is to correct the LaTeX code so that it accurately agrees with the table structure shown in the image.

---

### Information Provided:
- **Number of rows:** {rows}  
- **Number of columns:** {cols}  

### Initial LaTeX Prediction:
{initial_latex}

---

### Correction Instructions:
- Carefully review whether the initial LaTeX matches the table's layout in the image.
- Correct any structural issues, such as:
  - Missing or extra rows or columns
  - Incorrect use of \\multicolumn or \\multirow
  - Wrong column alignment (e.g., `{{ l c r }}`)
  - Missing or extra \\hline for visible borders
- Assume all cell contents are `"CELL"`. Your focus is on **table structure only**, not the actual text.
- Ensure the corrected LaTeX agrees with the provided number of rows and columns, that is, it contains **exactly {rows} rows and {cols} columns**, matching the visible layout.

---

### Corrected LaTeX:"""

entries = []
total_files = len(latex_preds)

for idx, (filename, initial_latex) in enumerate(latex_preds.items()):
    if filename not in rowcol_info:
        print(f"Warning: Skipping {filename} (no row/col info)")
        continue

    full_image_path = os.path.join(image_dir, filename)

    if not os.path.exists(full_image_path):
        print(f"Warning: Image not found: {full_image_path}")
        continue

    if filename not in groundtruth_latex:
        print(f"Warning: Groundtruth LaTeX not found for {filename}. Skipping...")
        continue

    rows, cols = rowcol_info[filename]["rows"], rowcol_info[filename]["cols"]
    with Image.open(full_image_path) as img:
        width, height = img.size

    prompt = create_prompt(rows, cols, initial_latex, groundtruth_latex[filename])

    entry = {
        "id": idx,
        "image": filename,
        "width": width,
        "height": height,
        "conversations": [
            {
                "from": "human",
                "value": prompt
            },
            {
                "from": "gpt",
                "value": groundtruth_latex[filename]
            }
        ]
    }
    entries.append(entry)

    if (idx + 1) % 1000 == 0 or idx + 1 == total_files:
        print(f"Processed {idx + 1}/{total_files} entries.")

with open(output_jsonl_path, "w") as out_file:
    for entry in entries:
        out_file.write(json.dumps(entry) + "\n")

print(f"✅ JSONL file created at: {output_jsonl_path}")
````

````python
import os
from tqdm import tqdm
import torch
import json

def load_rowcol_info(rowcol_path):
    """Load row and column information from file"""
    rowcol_info = {}
    with open(rowcol_path, "r") as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) >= 5:
                filename = parts[0]
                rows = int(parts[2].strip(","))
                cols = int(parts[4])
                rowcol_info[filename] = {"rows": rows, "cols": cols}
    return rowcol_info

def load_initial_latex_preds(latex_pred_path):
    """Load initial LaTeX predictions"""
    latex_preds = {}
    with open(latex_pred_path, "r") as f:
        for line in f:
            if '\t' not in line:
                continue
            filename, latex = line.strip().split('\t', 1)
            latex_preds[filename] = latex
    return latex_preds

def process_images_to_latex(src_text_file, image_dir, rowcol_path, initial_latex_path, output_file, model, tokenizer, generation_config):
    """Process images to generate LaTeX code using structured prompt with progress bar"""
    
    with open(src_text_file, "r") as file:
        image_ids = [line.strip() for line in file if line.strip()]
    
    rowcol_info = load_rowcol_info(rowcol_path)
    
    initial_latex_preds = {}
    if initial_latex_path:
        initial_latex_preds = load_initial_latex_preds(initial_latex_path)
    
    results_buffer = []
    
    for image_id in tqdm(image_ids, desc="Processing images", unit="image"):
        image_path = os.path.join(image_dir, image_id)
        
        if not os.path.exists(image_path):
            print(f"Warning: {image_path} not found, skipping.")
            continue
    
        if image_id not in rowcol_info:
            print(f"Warning: No row/col info for {image_id}, skipping.")
            continue
            
        rows, cols = rowcol_info[image_id]["rows"], rowcol_info[image_id]["cols"]
        
        initial_latex = initial_latex_preds.get(image_id, "No initial prediction available.")
        
        pixel_values = load_image(image_path, max_num=12).to(torch.bfloat16).cuda()
        
        prompt = f"""<image>

You are given a table image, an initial LaTeX prediction, and the correct number of rows and columns in the table. Your task is to correct the LaTeX code so that it accurately agrees with the table structure shown in the image.

---

### Information Provided:
- **Number of rows:** {rows}  
- **Number of columns:** {cols}  

### Initial LaTeX Prediction:
{initial_latex}

---

### Correction Instructions:
- Carefully review whether the initial LaTeX matches the table's layout in the image.
- Correct any structural issues, such as:
  - Missing or extra rows or columns
  - Incorrect use of \\multicolumn or \\multirow
  - Wrong column alignment (e.g., `{{ l c r }}`)
  - Missing or extra \\hline for visible borders
- Assume all cell contents are `"CELL"`. Your focus is on **table structure only**, not the actual text.
- Ensure the corrected LaTeX agrees with the provided number of rows and columns, that is, it contains **exactly {rows} rows and {cols} columns**, matching the visible layout.

---

### Corrected LaTeX:"""

        response = model.chat(tokenizer, pixel_values, prompt, generation_config)
        
        results_buffer.append({"image_filename": image_id, "latex_output": response})
        
        if len(results_buffer) >= 50:
            write_results_to_file(results_buffer, output_file)
    
    if results_buffer:
        write_results_to_file(results_buffer, output_file)

    print("LaTeX generation completed for all images.")

def write_results_to_file(results_buffer, output_file):
    """Write current buffer to file and clear it"""
    with open(output_file, 'a', encoding='utf-8') as f:
        for result in results_buffer:
            f.write(f"{result['image_filename']}\t{result['latex_output']}\n")
    results_buffer.clear()

src_text_file = "/tab2tex/tablev1.2/structure/src-test.txt"
image_dir = "/scratch/data/tab2tex/tablev1.2/images/square"
rowcol_path = "/scratch/data/tab2tex/table_counts_test_cell_based_with_filenames.txt"  # Row/col info for test set
initial_latex_path = "/unitable/FT_internvl_M2_inference_test.txt"
output_file = "/unitable/FT_internvl_M3_inference_test_CoTv2.txt"

process_images_to_latex(
    src_text_file=src_text_file,
    image_dir=image_dir,
    rowcol_path=rowcol_path,
    initial_latex_path=initial_latex_path,
    output_file=output_file,
    model=model,
    tokenizer=tokenizer,
    generation_config=generation_config
)
````
